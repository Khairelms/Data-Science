```python
# Working with Missing Data in Pandas
# Missing Data can occur when no information is provided for one or more
# items or for a whole unit. Missing Data is a very big problem in a real-life scenarios.
# Missing Data can also refer to as NA(Not Available) values in pandas.
# In DataFrame sometimes many datasets simply arrive with missing data,
# either because it exists and was not collected or it never existed.
# For Example, Suppose different users being surveyed may choose not to share their income,
# some users may choose not to share the address in this way many datasets went missing.

# Dataset 1 : https://raw.githubusercontent.com/yashy1626/ds_dataset/refs/heads/main/ufo.csv

# Dataset 2 : https://raw.githubusercontent.com/yashy1626/ds_dataset/refs/heads/main/sample11.csv
```

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

ufo = pd.read_csv('https://raw.githubusercontent.com/yashy1626/ds_dataset/refs/heads/main/ufo.csv')
ufo.head()
```

```python
ufo.info()
```

```python
ufo.isnull().sum()
```

```python
# remove missing values
# dropna()

ufo.dropna().shape          # if missing value is a lot, use fillna() to fill missing value
```

```python
# fillna()
# Fill NA/NaN values using the specified method
# for object dtype not for int or float

ufo['City'].fillna(value='CityNotMentioned',inplace=True)

ufo['Colors Reported'].fillna(value='ColorsNotFound',inplace=True)

ufo['Shape Reported'].fillna(value='ShapeNotFound',inplace=True)
```

```python
ufo.isnull().sum()
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/yashy1626/ds_dataset/refs/heads/main/sample11.csv')
data.info()
```

```python
data
```

```python
data.isnull().sum()
```

```python
# fillna() for fill missing value for int/flooat dtype using backward and forward fill

data['Marks'].fillna(method='bfill',inplace=True)

data['Percentage'].fillna(method='ffill',inplace=True)

```

```python
data
```

```python
# fill mising value by taking up a mean -- better approch

data = pd.read_csv('https://raw.githubusercontent.com/yashy1626/ds_dataset/refs/heads/main/sample11.csv')
data
```

```python
mean_val_marks = data['Marks'].mean()

data['Marks'].fillna(value=mean_val_marks,inplace=True)  # kalau x nk guna mean value boleh letak (value=0,inplace=True)

data
```

```python
mean_val_perc = data['Percentage'].mean()
mean_val_perc = round(mean_val_perc,1)   # kurangkan kpd 1 titik perpuluhan

data['Percentage'].fillna(value=mean_val_perc,inplace=True)

data
```

```python
# Remove Duplicate Records

# Handle Duplicate Data
# Dealing With Duplicate Data

# Create a DataFrame
data = {
      'StudentName': ['Mark', 'Ali', 'Bob', 'John', 'Johny', 'Mark'],
      'Score': [45, 65, 76, 44, 39, 45]
}
df = pd.DataFrame(data)
df
```

```python
# check for duplicate records

print(df.duplicated().sum())
```

```python
df[df.duplicated()]     # to print that duplicate records
```

```python
df[df['StudentName']== 'Mark']
```

```python
# dropping duplicates

df.drop_duplicates(inplace=True)

print(df.duplicated().sum())
```
