```python
# Pandas Data Science library

import numpy as np
import pandas as pd
print(pd.__version__)
```

```python
# numpy --> array()
arr = np.array([10,20,30,40,50])
print(type(arr))
print(arr)
```

```python
# Pandas Series
# Series is 1D Labelled Array

s = pd.Series([10,20,30,40,50])
print(type(s))
print(s)
print(s.shape)

# In Numpy --> Only Numbers
# In Series --> Number + Labels(index)
```

```python
info = pd.Series(['John',28,6700,'Klang'], index = ['Name','Age','Sal','City'])
print(info)
print()
print(info['Name'])       # indexing
print(info.get('City'))   # getting data values using get()
print(info.iloc[0])       # iloc -->index location

# Series is like a Dictionary + array combo
```

```python
# values property
# print only values

print(info.values)
```

```python
# Indexing Slicing And Using For Loop

s = pd.Series([100,200,300,400,500,600],index=['v1','v2','v3','v4','v5','v6'])
print(s)

# 100  200   300   400    500     600
# 0     1     2     3      4       5
# -6    -5    -4    -3     -2      -1
# v1     v2   v3     v4    v5      v6

# Indexing in Series can be done by using [] or iloc[] or loc[]
print('\nIndexing In Series')

print(s.iloc[0])    # for index numbers
print(s.loc['v5'])  # loc used for labels
print(s['v2'])      # [] used for labels
print(s.iloc[-1])

print('\nSlicing In Series')
print(s[1:4])         # 4 will be excluded --> stopIndex
print(s['v2':'v5'])   # if performing slicing using labels, the end label is included
```

```python
# print only 100,400 and 500

print(s['v1'],s['v4'],s['v5'])
print()
print(s[['v1','v4','v5']])        # for specific information s[['','','',]]
print()
print(s[['v1','v4','v5']].values) # just want the values output

# Conditional Selection
print('\nConditional Selection')
print(s[s%3==0])
```

```python
# Using for loop
for i in s:
  print(i)
```

```python
for k,v in s.items():
  print(k,v)
```

```python
for i in enumerate(s):
  print(i)
```
