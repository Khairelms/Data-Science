```python
# In Numpy Speed Is Better

import time

numbers = list(range(1,10000001))
total = 0

start_time = time.time()     # provide some start time
for n in numbers:
  total += n
end_time = time.time()     # provide some end time

print(total)
print(f"Total Time Taken : {end_time-start_time:.4f}")
```

```python
# copy() vs view()

arr = np.array([45,65,120,78,11,89])
print("Ori Array",arr)

# copy()
# create the copy of an array in a different memory location

copied_arr = arr.copy()
print ("Copied Array" ,copied_arr)

print()
# if you make any changes in copied_array it won't affect the original array

copied_arr[0] = 450
print("Copied Array",copied_arr)
print("Ori Array",arr)

print()
# if you make any changes in copied_array it won't affect the copied array

arr[3] = 780
print("Ori Array",arr)
print("Copied Array",copied_arr)
```

```python
# copy() vs view()

arr = np.array([45,65,120,78,11,89])
print("Ori Array",arr)

# view()
# is the representation of the original array
# it shares the same memory location

copied_arr = arr.view()
print ("Copied Array" ,copied_arr)

print()
# if you make any changes in copied_array it will affect the original array

copied_arr[0] = 450
print("Copied Array",copied_arr)
print("Ori Array",arr)
```

```python
# show me the proof
# view()
# is the representation of the original array
# it shares the same memory location

arr = np.array([45,65,120,78,11,89])
copied_arr = arr.view()

# share_memory() --> returns TRUE if arrays are sharing same memory location

print(np.shares_memory(arr,copied_arr))
```

```python
arr = np.array([10, 32, 30, 50, 20, 82, 91, 45])

# Find indices where value is 30

indices = np.where(arr == 30)

print(f"Indices where value is 30: {indices}")

# Find indices where values are even

even_indices = np.where(arr % 2 == 0)

print(f"Indices of even values: {even_indices}")

# Conditional selection

new_arr = np.where(arr > 40, arr * 2, arr)

print(f"Array after conditional selection: {new_arr}")

```

```python
# numpy.unique()

# return unique elements as a sorted array

a = np.unique([1,2,3,4,5,1,2,3,6,7,8,5,6,7,8,11,10,15,10,11,13,14,6])

print(a)

# Expected Output : [ 1  2  3  4  5  6  7  8 11 10 15 13 14]

# Return          : [ 1  2  3  4  5  6  7  8 10 11 13 14 15]  --> dublicate remove but by default it rearrange the number, we don't want it. need to sort back using next code
```

```python
arr = np.array([1,2,3,4,5,1,2,3,6,7,8,5,6,7,8,11,10,15,10,11,13,14,6])
uv , i = np.unique(arr,return_index=True)
print(uv) # unique_values
print(i) # index Numbers
```

```python
# argsort --> reorder based on original appearance
uv[np.argsort(i)]

# [ 1  2  3  4  5  6  7  8 10 11 13 14 15]
# [ 0  1  2  3  4  8  9 10 16 15 20 21 17]

# re order --> [ 0  1  2  3  4  8  9 10 15 16 17 20 21]
```

```python
arr = np.array([1,2,3,4,5,1,2,3,6,7,8,5,6,7,8,11,10,15,10,11,13,14,6])
uv , i = np.unique(arr,return_index=True)
uv[np.argsort(i)]
```

```python
# Numpy Broadcasting

arr1 = np.array([5,8,10,12])
arr2 = np.array([3,7,6,5])

print(arr1+arr2)
print(arr1*arr2)
```

```python
# Broadcasting is an error
# and this error will shown up
# when we perform arithmetic operation b/w uncompatible shapes

arr1 = np.array([5,8,10,12])
arr2 = np.array([3,7])

print(arr1+arr2)
```

```python
# General Rules To Avoid Broadcasting

# a.) both array dimensions should be equal
# b.) if any of them is 1

arr1 = np.array([5,8,10,12])
arr2 = np.array([3])

print(arr1+arr2)
```
