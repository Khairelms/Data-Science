```python
plt.figure(figsize=(18,5))

x1 = data[data['churn'] == 0].index
y1 = data[data['churn'] == 0]['total day minutes']

x2 = data[data['churn'] == 1].index
y2 = data[data['churn'] == 1]['total day minutes']

plt.plot(x1, y1, color='green', linewidth=1, linestyle='solid', label='Active User')
plt.plot(x2, y2, color='b', linewidth=1, linestyle='solid', label='Churn User')
plt.xlabel("User")
plt.ylabel("Total Day Minutes")
plt.legend()
plt.title("Total Day Minutes by Churn Status")
plt.grid(True,color='lightblue')
plt.show()
```

```python
# Import numpy for numerical operations (used to create x-axis positions)
import numpy as np

# Import matplotlib for plotting graphs
import matplotlib.pyplot as plt


# -------------------------------
# 1. Separate the data
# -------------------------------

# Create a DataFrame with only NON-CHURN users (churn = 0)
non_churn = data[data['churn'] == 0]

# Create a DataFrame with only CHURN users (churn = 1)
churn = data[data['churn'] == 1]


# -------------------------------
# 2. Columns we want to compare
# -------------------------------

# List of columns that store minute information
infoper_minutes = [
    'total day minutes',
    'total eve minutes',
    'total night minutes'
]


# -------------------------------
# 3. Calculate average (mean) minutes
# -------------------------------

# Calculate average minutes for NON-CHURN users
non_churn_mean = non_churn[infoper_minutes].mean()

# Calculate average minutes for CHURN users
churn_mean = churn[infoper_minutes].mean()


# -------------------------------
# 4. Prepare x-axis positions
# -------------------------------

# Create numeric positions for each bar (0, 1, 2)
x = np.arange(len(infoper_minutes))

# Width of each bar
width = 0.35


# -------------------------------
# 5. Draw the bar chart
# -------------------------------

# Plot bars for NON-CHURN users (shifted left)
plt.bar(x - width/2, non_churn_mean, width, label='Non-Churn')

# Plot bars for CHURN users (shifted right)
plt.bar(x + width/2, churn_mean, width, label='Churn')


# -------------------------------
# 6. Add labels and title
# -------------------------------

# Set labels for x-axis using column names
plt.xticks(x, infoper_minutes, rotation=15)

# Label for x-axis
plt.xlabel('Minute Type')

# Label for y-axis
plt.ylabel('Average Minutes')

# Title of the graph
plt.title('Average Usage: Churn vs Non-Churn')

# Show legend to differentiate bars
plt.legend()


# -------------------------------
# 7. Display the graph
# -------------------------------

plt.show()
```
